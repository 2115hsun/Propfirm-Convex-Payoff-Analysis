# Funded account model: roadmap

How to use this file:
- Tick a box by changing `[ ]` to `[x]`. In VS Code with the "Markdown All in One" extension, Alt+C toggles the box under the cursor.
- Work top to bottom in **Next up**. Pull items from **Backlog** into Next up when there's room.
- An item is done only when its **check** passes. Note the result in the **Change log** at the bottom.
- Commit after each item (`git commit -m "A3: commission in coins"`) so every number can be traced to a version.
